```python
#MAX AMONG 3 NUMBERS
a=int(input("enter a:"))
b=int(input("enter b:"))
c=int(input("enter c:"))
if a>b and a>c :
  print("a is greater")
elif b>c and b>a :
  print("b is greater")
else :
  print("c is greater")
```

```python
#LEAP YEAR
a=int(input("enter year :"))
if a%100!=0 and a%4==0 :
  print("year is leap")
else :
  print("not a leap")
```

```python
#EXAMPLE CODE 0N MATCH CASE
day=int(input())
match day :
  case 1 :
    print("monday")
  case 2 :
    print("tuesday")
  case 3:
     print("wednesday")
  case 4 :
    print("friday")
  case 5 :
    print("saturday")
  case _:
    print("invalid")
```

```python
#SUM OF DIGITS OF A GIVEN NUMBER
a=int(input())
sum=0
while (a!=0):
  res=a%10
  sum=sum+res
  a=a//10
print(sum)
```

```python
#SUM OF SQUARES OF DIGITS IN NUMBER
a=int(input())
sum=0
while(a!=0) :
  res=a%10
  res=res*res
  sum=sum+res
  a=a//10
print(sum)
```

```python
#AMSTRONG NUMBER
a=int(input())
sum=0
while(a!=0) :
  res=a%10
  res=res*res*res
  sum=sum+res
  a=a//10
print(sum)
```

```python
#PRIME NUMBER OR NOT
a=int(input())
for i in range(2,a):
  if a%i==0 :
    print("not a prime")
    break
else :
    print("prime")
```

```python
#PRINT PRIME NUMBERS IN GIVEN LIMIT
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if i%j==0 :
      break
  else :
    print(i)
```
