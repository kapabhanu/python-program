```python
#FUNCTIONS IN PYTHON
#program that counts vowels
a="bhanu yasaswiniIII"
count = 0
for i in a:
  if i.lower() in 'aeiou' or i.upper() in'AEIOU':
    count=count+1
print(count)
```

```python
#to reverse a number
num=int(input())
def reverse(num):
  rev = 0
  while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
  return rev
print(reverse(num))
```
