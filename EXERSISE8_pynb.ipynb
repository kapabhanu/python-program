```python
#python program to find no of duplicates from a list
l1 = [int(i) for i in input().split()]
s1= set(l1)
print(len(l1)-len(s1))
```

```python
#python program to print non repeated elements
l1 = [int(i) for i in input().split()]
nr = []
for i in l1:
  if l1.count(i) == 1:
    nr.append(i)
print(nr)
```

```python
#program that prints missing numbers between limit 1 to n
n = int(input())
l1 = [int(x) for x in input().split()]
s1 = set(l1)
s2= []
for i in range(1, n + 1):
  if i not in s1:
    s2.append(i)
print(s2)
```

```python
#set methods
a={1,2,3,4}
a.add(5) #add
print(a)
#update
a.update([4,5,6])
print(a)
#remove
a.remove(3)
print(a)
#discard
a.discard(99)
print(a)
b={2,3,4,7}
print(a.difference(b))#difference
print(a.union(b))#union
print(a.intersection(b))#intersection
print(a.symmetric_difference(b))#symmetric difference
print(a.pop())#pop
print(a.clear())#clear
```
