```python
#OPERATIONS ON LIST
      #1.CONCATENATION
a1=[16,38,47,'joshitha']
a2=[9,21,'bhanu']
print(a1+a2)
      #2.REPETITION
print(a1*3)
      #3.INDEXING
print(a1[3])
print(a2[2])
     #4.SLICING
print(a1[0:3])
print(a2[0:2])
     #5.MEMBERSHIP OPERATOR
if 21 in a2 :
  print("yes")
else :
  print("no")
     #6.ITERATION
for i in a2:
  print(i)
for i in range(len(a2)):
  print(i)
  print(a2[i])
```

```python
#BUILT IN FUNCTIONS
b1=[45,85,36,4,81,91,92,6]
#1.len
print(len(b1))
#2.sum
print(sum(b1))
#3.min
print(min(b1))
#4.max
print(max(b1))
#5.reversed
print(list(reversed(b1)))
#6.sort
print(sorted(b1))
```

```python
#BUILT IN METHODS
a1=[10,11,12,13]
b2=[15,94,1,65,65,11]
#1.append
b2.append("bhanu")
print(b2)
#2.insert
b2.insert(0,"hello")
print(b2)
#3.extend
b2.extend(a1)
print(b2)
#4.pop
b2.pop()
print(b2)
#5.count
print(b2.count(11))
#6.index
print(b2.index(65))
```

```python
#PROGRAM THAT PRINTS SUM OF ELEMENTS WITHOUT USING SUM
l1=[45,65,8,59,94,54]
s=0
for i in l1:
  s=s+i
print(s)
```

```python
#PRINTS NUMBERS IN ODD INDEX
for i in range(len(l1)):
  if(i%2!=0):
    print(l1[i])
```
